# Lab 2: 02 Procesamiento de Ortofotos SAF (IDE Chile)

Este notebook se enfoca en integrar las ortofotos de alta resolución proporcionadas por el SAF (Servicio Aerofotogramétrico) a través de IDE Chile.

```python
import rasterio
from rasterio.plot import show
import matplotlib.pyplot as plt
import os

# Ruta a la ortofoto descargada desde IDE Chile
ortofoto_path = "../data/raw/ide_chile/ortofoto_2023.tif"

if os.path.exists(ortofoto_path):
    with rasterio.open(ortofoto_path) as src:
        print(f"Dimensiones: {src.width}x{src.height}")
        print(f"CRS: {src.crs}")
        show(src)
else:
    print("Por favor, descarga la ortofoto desde https://www.ide.cl/ y colócala en data/raw/ide_chile/")
```
